```python
# ============================================================
# CELDA 1
# Importar librerías
# ============================================================

import pandas as pd
import re
import unicodedata

from pathlib import Path
from IPython.display import display

pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", 250)
```

```python
# ============================================================
# CELDA 2
# Cargar la matriz de estudios incluidos
# ============================================================

archivo_csv = "estudios_primarios.csv"

df = pd.read_csv(
    archivo_csv,
    sep=";",
    encoding="utf-8-sig"
)

print("Total de estudios incluidos:", len(df))
print("Número de columnas:", len(df.columns))
```

```python
# ============================================================
# CELDA 3
# Selección de evidencia directa para RQ4
#
# Se consideran:
#   - Principal
#   - Sí
#
# Se excluyen de la síntesis principal:
#   - Parcial
#   - No aborda
#   - No
#   - Parcial (no seguridad)
# ============================================================

print("Distribución completa de RQ4:")
print(df["RQ4"].value_counts(dropna=False))

df_rq4 = df[
    df["RQ4"].isin([
        "Principal",
        "Sí"
    ])
].copy()

print(
    "\nEstudios directamente designados a RQ4:",
    len(df_rq4)
)
```

```python
# ============================================================
# CELDA 4
# Campos utilizados para responder RQ4
# ============================================================

columnas_rq4 = [
    "ID",
    "Título",
    "Mecanismo de refinamiento",
    "Tipo de retroalimentación",
    "Mecanismo de verificación",
    "RQ4"
]

display(
    df_rq4[columnas_rq4]
    .sort_values("ID")
)
```

```python
# ============================================================
# CELDA 5
# Leer bibliografía LaTeX
# ============================================================

archivo_bib = "bibliografia.md"

texto_bib = Path(
    archivo_bib
).read_text(
    encoding="utf-8",
    errors="replace"
)

patron_bibitem = re.compile(
    r"\\bibitem\{([^}]+)\}\s*"
    r"(.*?)"
    r"(?=\\bibitem\{|\Z)",
    flags=re.DOTALL
)

referencias = []

for clave, bloque in patron_bibitem.findall(texto_bib):

    titulo_match = re.search(
        r"\\textit\{(.*?)\};",
        bloque,
        flags=re.DOTALL
    )

    if titulo_match:

        referencias.append({
            "BibKey": clave,
            "Título_bib": titulo_match.group(1).strip()
        })

bibliografia = pd.DataFrame(referencias)

print(
    "Referencias encontradas:",
    len(bibliografia)
)

display(bibliografia.head())
```

```python
# ============================================================
# CELDA 6
# Normalizar títulos para realizar el match
# ============================================================

def normalizar_titulo(texto):

    texto = str(texto)

    texto = unicodedata.normalize(
        "NFKD",
        texto
    )

    texto = "".join(
        c for c in texto
        if not unicodedata.combining(c)
    )

    texto = (
        texto
        .lower()
        .replace("", "'")
        .replace("–", "-")
        .replace("—", "-")
    )

    texto = re.sub(
        r"[^a-z0-9]+",
        " ",
        texto
    )

    texto = re.sub(
        r"\s+",
        " ",
        texto
    )

    return texto.strip()
```

```python
# ============================================================
# CELDA 7
# Match entre los estudios de RQ4 y la bibliografía
# ============================================================

bibliografia["titulo_normalizado"] = (
    bibliografia["Título_bib"]
    .apply(normalizar_titulo)
)

df_rq4["titulo_normalizado"] = (
    df_rq4["Título"]
    .apply(normalizar_titulo)
)

diccionario_bib = dict(
    zip(
        bibliografia["titulo_normalizado"],
        bibliografia["BibKey"]
    )
)

df_rq4["BibKey"] = (
    df_rq4["titulo_normalizado"]
    .map(diccionario_bib)
)

sin_match = df_rq4[
    df_rq4["BibKey"].isna()
]

print(
    "Estudios sin correspondencia bibliográfica:",
    len(sin_match)
)

if len(sin_match) == 0:
    print(
        "✓ Los 10 estudios tienen correspondencia bibliográfica."
    )
else:
    display(
        sin_match[
            ["ID", "Título"]
        ]
    )
```

### Tabla 1 — Mecanismos de refinamiento
En esta tabla las categorías son no excluyentes, porque un mismo enfoque puede combinar corrección, regeneración y refinamiento iterativo.\
— Definir categorías

```python
# ============================================================
# CELDA 8
# Categorías de mecanismos de refinamiento
#
# IMPORTANTE:
# Estas categorías NO son mutuamente excluyentes.
# Un mismo estudio puede implementar varios mecanismos.
# ============================================================

categorias_refinamiento = {

    "Corrección o reparación posterior a la verificación": [
        10, 21, 28, 29, 60, 66
    ],

    "Regeneración o reintento automático": [
        21, 29
    ],

    "Refinamiento iterativo en bucle cerrado": [
        10, 21, 28, 29, 47, 60, 66, 77
    ],

    "Human-in-the-loop": [
        28, 66
    ],

    "Feedback de verificación hacia la arquitectura o configuración": [
        29, 47, 60, 66
    ],

    "Refinamiento adaptativo del conocimiento o recuperación": [
        49, 67
    ]
}
```

```python
# ============================================================
# CELDA 9
# Descripción sintética
# ============================================================

descripcion_refinamiento = {

    "Corrección o reparación posterior a la verificación":
        "Los errores, inconsistencias, smells o violaciones "
        "detectadas activan modificaciones sobre el artefacto.",

    "Regeneración o reintento automático":
        "Una salida que no satisface la verificación se vuelve "
        "a generar, reparar o evaluar.",

    "Refinamiento iterativo en bucle cerrado":
        "El ciclo generar/evaluar/corregir se repite utilizando "
        "los resultados de la iteración anterior.",

    "Human-in-the-loop":
        "Expertos o desarrolladores validan resultados y sus "
        "decisiones modifican requisitos o refactorizaciones.",

    "Feedback de verificación hacia la arquitectura o configuración":
        "Los resultados de verificación o rendimiento modifican "
        "el modelo, refactoring o configuración arquitectónica.",

    "Refinamiento adaptativo del conocimiento o recuperación":
        "La relevancia de la consulta o la utilidad del contexto "
        "ajustan dinámicamente el conocimiento recuperado."
}
```

```python
# ============================================================
# CELDA 10
# Tabla de mecanismos de refinamiento
# ============================================================

filas = []

for categoria, ids in categorias_refinamiento.items():

    estudios = (
        df_rq4[
            df_rq4["ID"].isin(ids)
        ]
        .sort_values("ID")
    )

    citas = (
        "\\cite{"
        + ",".join(
            estudios["BibKey"].tolist()
        )
        + "}"
    )

    filas.append({
        "Mecanismo de refinamiento": categoria,
        "Descripción": descripcion_refinamiento[categoria],
        "Estudios designados": citas,
        "Total": len(estudios)
    })


tabla_rq4_refinamiento = pd.DataFrame(filas)

display(tabla_rq4_refinamiento)
```

### Tabla 2 — Tipos de retroalimentación
Para esta segunda tabla sí podemos asignar a cada estudio el tipo dominante de retroalimentación, evitando duplicados.\
— Categorizar la retroalimentación

```python
# ============================================================
# CELDA 11
# Categorías de retroalimentación
#
# Aquí cada estudio se asigna a una categoría dominante.
# ============================================================

categorias_feedback = {

    "Resultados de validación o ejecución": [
        10, 21, 60
    ],

    "Evaluación de expertos o desarrolladores": [
        28, 66
    ],

    "Telemetría, rendimiento y cumplimiento de SLA": [
        47, 77
    ],

    "Guardrails y autoevaluación del modelo": [
        29, 49
    ],

    "Relevancia de consulta y estado de la base de conocimiento": [
        67
    ]
}
```

```python
# ============================================================
# CELDA 12
# Destino de la retroalimentación
# ============================================================

destino_feedback = {

    "Resultados de validación o ejecución":
        "Corrección o regeneración de ontologías, modelos "
        "y artefactos generados.",

    "Evaluación de expertos o desarrolladores":
        "Revisión de requisitos, casos reutilizados "
        "y decisiones de refactoring.",

    "Telemetría, rendimiento y cumplimiento de SLA":
        "Optimización de configuración, recursos, "
        "políticas y decisiones posteriores.",

    "Guardrails y autoevaluación del modelo":
        "Reparación de candidatos arquitectónicos o decisión "
        "sobre el uso de conocimiento recuperado.",

    "Relevancia de consulta y estado de la base de conocimiento":
        "Restricción y actualización dinámica del espacio "
        "de conocimiento utilizado en la recuperación."
}
```

```python
# ============================================================
# CELDA 13
# Construir tabla de retroalimentación
# ============================================================

filas = []

for tipo, ids in categorias_feedback.items():

    estudios = (
        df_rq4[
            df_rq4["ID"].isin(ids)
        ]
        .sort_values("ID")
    )

    citas = (
        "\\cite{"
        + ",".join(
            estudios["BibKey"].tolist()
        )
        + "}"
    )

    filas.append({
        "Tipo de retroalimentación": tipo,
        "Destino / efecto": destino_feedback[tipo],
        "Estudios designados": citas,
        "Total": len(estudios)
    })


tabla_rq4_feedback = pd.DataFrame(filas)

display(tabla_rq4_feedback)
```

```python
# ============================================================
# CELDA 14
# Control de consistencia
# ============================================================

ids_feedback = [
    estudio
    for ids in categorias_feedback.values()
    for estudio in ids
]

ids_rq4 = set(
    df_rq4["ID"]
)

ids_feedback_unicos = set(
    ids_feedback
)

duplicados = [
    x
    for x in ids_feedback
    if ids_feedback.count(x) > 1
]

print(
    "Estudios RQ4:",
    len(ids_rq4)
)

print(
    "Estudios categorizados:",
    len(ids_feedback_unicos)
)

print(
    "Faltantes:",
    sorted(
        ids_rq4 - ids_feedback_unicos
    )
)

print(
    "Adicionales:",
    sorted(
        ids_feedback_unicos - ids_rq4
    )
)

print(
    "Duplicados:",
    sorted(set(duplicados))
)
```

```python
# ============================================================
# CELDA 16
# Escapar caracteres reservados de LaTeX
# ============================================================

def escapar_latex(texto):

    texto = str(texto)

    reemplazos = {
        "&": r"\&",
        "%": r"\%",
        "#": r"\#",
        "_": r"\_"
    }

    for original, nuevo in reemplazos.items():
        texto = texto.replace(
            original,
            nuevo
        )

    return texto
```

```python
# ============================================================
# CELDA 17
# Tabla LaTeX de mecanismos de refinamiento
# ============================================================

def generar_latex_refinamiento(tabla):

    lineas = []

    lineas.append(r"\begin{table}[H]")

    lineas.append(
        r"\caption{Mecanismos de refinamiento identificados "
        r"en los estudios que responden a RQ4."
        r"\label{tab:rq4-refinamiento}}"
    )

    lineas.append(
        r"%\isPreprints{\centering}{%"
    )

    lineas.append(
        r"\begin{adjustwidth}{-\extralength}{0cm}"
    )

    lineas.append(r"%}")

    lineas.append(
        r"%\isPreprints{\begin{tabularx}"
        r"{\textwidth}{CCCC}}{%"
    )

    lineas.append(
        r"\begin{tabularx}{\fulllength}"
        r"{p{0.25\fulllength} "
        r"p{0.30\fulllength} "
        r"X "
        r"p{0.06\fulllength}}"
    )

    lineas.append(r"%}")
    lineas.append(r"\toprule")

    lineas.append(
        r"\textbf{Mecanismo de refinamiento} & "
        r"\textbf{Descripción} & "
        r"\textbf{Estudios designados} & "
        r"\textbf{Total} \\"
    )

    lineas.append(r"\midrule")
    lineas.append("")

    for _, fila in tabla.iterrows():

        mecanismo = escapar_latex(
            fila["Mecanismo de refinamiento"]
        )

        descripcion = escapar_latex(
            fila["Descripción"]
        )

        citas = fila["Estudios designados"]

        total = fila["Total"]

        lineas.append(
            f"{mecanismo} & "
            f"{descripcion} & "
            f"{citas} & "
            f"{total} \\\\"
        )

        lineas.append("")

    lineas.append(r"\bottomrule")
    lineas.append(r"\end{tabularx}")

    lineas.append(
        r"%\isPreprints{}{%"
    )

    lineas.append(
        r"\end{adjustwidth}"
    )

    lineas.append(r"%}")
    lineas.append(r"\end{table}")

    return "\n".join(lineas)


latex_refinamiento = generar_latex_refinamiento(
    tabla_rq4_refinamiento
)

print(latex_refinamiento)
```

```python
# ============================================================
# CELDA 18
# Tabla LaTeX de tipos de retroalimentación
# ============================================================

def generar_latex_feedback(tabla):

    lineas = []

    lineas.append(r"\begin{table}[H]")

    lineas.append(
        r"\caption{Tipos de retroalimentación y su efecto "
        r"en los estudios que responden a RQ4."
        r"\label{tab:rq4-feedback}}"
    )

    lineas.append(
        r"%\isPreprints{\centering}{%"
    )

    lineas.append(
        r"\begin{adjustwidth}{-\extralength}{0cm}"
    )

    lineas.append(r"%}")

    lineas.append(
        r"%\isPreprints{\begin{tabularx}"
        r"{\textwidth}{CCCC}}{%"
    )

    lineas.append(
        r"\begin{tabularx}{\fulllength}"
        r"{p{0.25\fulllength} "
        r"p{0.31\fulllength} "
        r"X "
        r"p{0.06\fulllength}}"
    )

    lineas.append(r"%}")
    lineas.append(r"\toprule")

    lineas.append(
        r"\textbf{Tipo de retroalimentación} & "
        r"\textbf{Destino / efecto} & "
        r"\textbf{Estudios designados} & "
        r"\textbf{Total} \\"
    )

    lineas.append(r"\midrule")
    lineas.append("")

    for _, fila in tabla.iterrows():

        tipo = escapar_latex(
            fila["Tipo de retroalimentación"]
        )

        destino = escapar_latex(
            fila["Destino / efecto"]
        )

        citas = fila["Estudios designados"]

        total = fila["Total"]

        lineas.append(
            f"{tipo} & "
            f"{destino} & "
            f"{citas} & "
            f"{total} \\\\"
        )

        lineas.append("")

    lineas.append(r"\midrule")

    lineas.append(
        r"\textbf{Total} & & & \textbf{10} \\"
    )

    lineas.append(r"\bottomrule")
    lineas.append(r"\end{tabularx}")

    lineas.append(
        r"%\isPreprints{}{%"
    )

    lineas.append(
        r"\end{adjustwidth}"
    )

    lineas.append(r"%}")
    lineas.append(r"\end{table}")

    return "\n".join(lineas)


latex_feedback = generar_latex_feedback(
    tabla_rq4_feedback
)

print(latex_feedback)
```

```python
# ============================================================
# CELDA 19
# Guardar resultados
# ============================================================

with open(
    "tabla_rq4_refinamiento.tex",
    "w",
    encoding="utf-8"
) as archivo:

    archivo.write(latex_refinamiento)


with open(
    "tabla_rq4_feedback.tex",
    "w",
    encoding="utf-8"
) as archivo:

    archivo.write(latex_feedback)


print("Archivos generados:")
print(" - tabla_rq4_refinamiento.tex")
print(" - tabla_rq4_feedback.tex")
```

```python
# ============================================================
# CELDA 20
# Guardar evidencia detallada de RQ4
# ============================================================

# Categorías múltiples de refinamiento
def categorias_de_estudio(id_estudio):

    categorias = []

    for categoria, ids in categorias_refinamiento.items():

        if id_estudio in ids:
            categorias.append(categoria)

    return "; ".join(categorias)


# Categoría dominante de feedback
feedback_por_id = {}

for categoria, ids in categorias_feedback.items():

    for id_estudio in ids:
        feedback_por_id[id_estudio] = categoria


df_rq4["Categorías de refinamiento"] = (
    df_rq4["ID"]
    .apply(categorias_de_estudio)
)

df_rq4["Categoría de retroalimentación"] = (
    df_rq4["ID"]
    .map(feedback_por_id)
)


trazabilidad_rq4 = df_rq4[
    [
        "ID",
        "BibKey",
        "Título",
        "Mecanismo de refinamiento",
        "Categorías de refinamiento",
        "Tipo de retroalimentación",
        "Categoría de retroalimentación",
        "Trazabilidad",
        "Explicabilidad"
    ]
].sort_values("ID")


display(trazabilidad_rq4)

trazabilidad_rq4.to_excel(
    "trazabilidad_RQ4.xlsx",
    index=False
)

print(
    "Archivo generado: trazabilidad_RQ4.xlsx"
)
```
